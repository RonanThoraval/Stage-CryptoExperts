```python
def KeyGen(G,d):
    return [G.random_element() for _ in range(d+1)]
```

```python
def Commit(P,gp):
    d = P.degree()
    return sum([P[i]*gp[i] for i in range(d+1)])
```

```python
def Proof(u,P,gp):
    v = P(u)
    d = P.degree()
    D = (d+1)/2
    print("D = ", D," and |gp| = ", len(gp))
    L = sum([P[i]*gp[i+D] for i in range(D)])
    R = sum([P[i+D]*gp[i] for i in range(D)])
    vL = sum([P[i]*u**i for i in range(D)])
    vR = sum([P[i+D]*u**i for i in range(D)])
    return v,L,R,vL,vR
```

```python
def Verif(d,u,v,vL,vR):
    D = (d+1)/2
    return v ==(vL + vR*u**D)
```

```python
def CommitVerifior(r,R,L,comP):
    return r*L +comP +(r**(-1))*R
```
