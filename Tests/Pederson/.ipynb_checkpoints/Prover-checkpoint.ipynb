```python
q = random_prime(100)
Fq = GF(q)
FqX.<X> = PolynomialRing(Fq)
```

```python
n = randint(1,5)
d = 2**n-1
```

```python
P = FqX.random_element(d)
```

```python
p = random_prime(100)
Fp = GF(p)
G = EllipticCurve(Fp, [randint(0,p-1) for _ in range(5)])
qtmp = G.order()
while(qtmp != q):
    G = EllipticCurve(Fp, [randint(0,p-1) for _ in range(5)])
    qtmp = G.order()
```

```python
def Parameters():
    return d,q,G
```

```python
gp = [G.random_element() for _ in range(d+1)]
def KeyGen():
    return gp
```

```python
def Commit():
    return sum([P[i]*gp[i] for i in range(d+1)])
```

```python
def Proof(u):
    v = P(u)
    D = (d+1)/2
    L = sum([P[i]*gp[i+D] for i in range(D)])
    R = sum([P[i+D]*gp[i] for i in range(D)])
    vL = sum([P[i]*u**i for i in range(D)])
    vR = sum([P[i+D]*u**i for i in range(D)])
    return v,L,R,vL,vR
```
