```python
def KeyGen(Fq,G,n):
    
    s = Fq.random_element()
    g_tmp = G.gen(0)
    gp = [0]*n
    
    for i in range(n):
        gp[i] = g_tmp
        g_tmp = s*g_tmp
    return gp,s
```

```python
def commit(P,gp):
    return sum([P[i]*gp[i] for i in range(P.degree()+1)])
```

```python
def Proof(P,a,b,gp):
    Q,_ = (P-b).quo_rem(X-a)
    return commit(Q,gp)
```

```python
def Verif(a,b,comP,comQ,gp,q,p):    
    Ptmp = comP-b*gp[0]
    print("Ptmp = ", Ptmp)
    nP = Ptmp.order()
    print("nP = ", nP)
    kP = GF(nP)(p).multiplicative_order()
    
    nQ = comQ.order()
    kQ = GF(nQ)(p).multiplicative_order()
    
    print("first = ", Ptmp.tate_pairing(gp[0],nP,kP))
    
    print("second = ", comQ.tate_pairing(gp[1]-a*gp[0],nQ,kQ))
    
    return Ptmp.tate_pairing(gp[0],nP,kP)==comQ.tate_pairing(gp[1]-a*gp[0],nQ,kQ)
```

```python
Verif(r,P(r),comP,comQ,gp,q,p)
```
