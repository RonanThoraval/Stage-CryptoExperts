```python
nb_gates = randint(10,100000)

nb_inputs =randint(2,nb_gates//4)

r = []

for i in range(nb_gates):
    gate = randint(0,12)
    #while gate == 3:
        #gate = randint(0,6)
    #gate = 0
    inp1 = randint(0,nb_inputs-1+i)
    inp2 = randint(0,nb_inputs-1+i)
    r.append((gate,inp1,inp2))
```

```python
word_size = 2*randint(2,14)
test_function = random_function(r,word_size)
```

```python
C_BitW = fun_to_circuit_BitWise(test_function,nb_inputs,word_size,True)    
C_Arith,dict_proof = fun_to_circuit_Arithmetic(C_BitW,word_size,True)
```

```python
C_Arith.node_counter/C_BitW.node_counter
```

```python
inp = [randint(0,2^word_size-1) for _ in range(nb_inputs)]
Trace_BitW = C_BitW.trace(inp)
C_BitW.evaluate(inp)
```

```python
w = witness(Trace_BitW,dict_proof,word_size)
inp_Arith = input_Arith(inp,w)
C_Arith.evaluate(inp_Arith)
```

```python
cpt = 1
while(C_Arith.evaluate(inp_Arith)==C_BitW.evaluate(inp)+[True]):
    nb_gates = randint(10,100000)

    nb_inputs =randint(2,nb_gates//4)
    r = []

    for i in range(nb_gates):
        gate = randint(0,12)
        #while gate == 3:
            #gate = randint(0,12)
        #gate = 0
        inp1 = randint(-1,nb_inputs-1+i)
        inp2 = randint(-1,nb_inputs-1+i)
            
        r.append((gate,inp1,inp2))
    
    word_size = 2*randint(2,14)
    test_function = random_function(r,word_size)
    
    C_BitW = fun_to_circuit_BitWise(test_function,nb_inputs,word_size,True)    
    C_Arith,dict_proof = fun_to_circuit_Arithmetic(C_BitW,word_size,True)
    
    inp = [randint(0,2^word_size-1) for _ in range(nb_inputs)]
    Trace_BitW = C_BitW.trace(inp)
    
    w = witness(Trace_BitW,dict_proof,word_size)
    inp_Arith = input_Arith(inp,w)
    sys.stdout.write("\r")
    sys.stdout.write("%d"%cpt)
    cpt += 1
```
