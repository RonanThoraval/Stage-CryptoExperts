```python
def witness(Trace_BitW,dict_proof,word_size=64):
    #print("WITNESS :")
    
    w = ()
    
    dict_decomp = {}
    dict_AND_XOR_OR = {}
    dict_NOT = {}
    dict_SHL = {}
    dict_SHR = {}
    dict_ADD = {}
    dict_SUB = {}
    dict_MUL = {}
    dict_DIV_MOD = {}
    
    var_calc = {}
    
    for node in Trace_BitW:
        
        if node.is_INPUT():
            
            inp = Trace_BitW[node]
            w += decomp(inp,word_size,dict_decomp)
            
        elif not node.is_CONST():
             if dict_proof[node]:
            
                a,b = get_inputs_nodes(node,var_calc)

                #print(a,b,end = ' : ')

                if node.is_AND() or node.is_XOR() or node.is_OR():

                    w += find_witness_AND_XOR_OR(node,a,b,word_size,Trace_BitW,dict_decomp,dict_AND_XOR_OR,var_calc)

                elif node.is_NOT():

                    w += find_witness_NOT(node,a,word_size,Trace_BitW,dict_decomp,dict_NOT,dict_AND_XOR_OR,var_calc)

                elif node.is_SHL():
                    b = node.operation.shift

                    w += find_witness_SHL(node,a,b,word_size,Trace_BitW,dict_decomp,dict_SHL,dict_MUL,var_calc)


                elif node.is_SHR():
                    b = node.operation.shift

                    w += find_witness_SHR(node,a,b,word_size,Trace_BitW,dict_decomp,dict_SHR,var_calc)

                #elif node.is_ROL():

                #elif node.is_ROR():

                elif node.is_ADD():

                    w += find_witness_ADD(node,a,b,word_size,Trace_BitW,dict_decomp,dict_ADD,var_calc)

                elif node.is_SUB():

                    w += find_witness_SUB(node,a,b,word_size,Trace_BitW,dict_decomp,dict_SUB,var_calc)

                elif node.is_MUL():

                    w += find_witness_MUL(node,a,b,word_size,Trace_BitW,dict_decomp,dict_MUL,var_calc)

                elif node.is_DIV() or node.is_MOD():
                    w += find_witness_DIV_MOD(node,a,b,word_size,Trace_BitW,dict_decomp,dict_DIV_MOD,var_calc)

                #elif node.is_NEG():
        
    return tuple(w)
```
