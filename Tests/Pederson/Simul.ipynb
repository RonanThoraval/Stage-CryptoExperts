```python
p = random_prime(100)
print("p = ",p)
Fp = GF(p)
G = EllipticCurve(Fp, [randint(0,p-1) for _ in range(5)])
q=G.order()
while(not(is_prime(q))):
    G = EllipticCurve(Fp, [randint(0,p-1) for _ in range(5)])
    q=G.order()
print("q = ",q)
Fq = GF(q)
```
